```python
import rdkit
```

```python
hbd_chuckles = "N[C@@H](Cc1c[nH]cn1)C(=O)|N[C@@H](Cc1ccc(O)cc1)C(=O)|N[C@@H](CCSC)C(=O)|N(C)[C@@H](CCCCN)C(=O)|N[C@H](CCCCN)C(=O)|N[C@@H](C(C)C)C(=O)|N[C@@H](CCC(=O)N)C(=O)|N[C@@H]([C@@H](C)O)C(=O)|N[C@H](CO)C(=O)|NCC(=O)|NCC(=O)|N[C@@H](C1CN(C(=O)N(CCC#N)C2CC2)C1)C(=O)|N[C@@H](Cc1c[nH]cn1)C(=O)|N[C@@H](CN[C@@H]CN[C@@H]C(=O)N)C(=O)|NCC(=O)|N[C@@H]N[C@@H](Cc1c[nH]c2ccccc12)C(=O)"
amino_acids = hbd_chuckles.split('|')

peptide = ''.join(amino_acids)
```

```python
from rdkit import Chem
from rdkit.Chem import Draw
from rdkit.Chem.rdMolDescriptors import CalcNumHBD

# Convert SMILES strings to RDKit molecule objects
amino_acids.append(peptide)
molecules = [Chem.MolFromSmiles(smiles) for smiles in amino_acids]

# Calculate the number of HBD for each molecule
hbd_counts = [CalcNumHBD(mol) for mol in molecules]

# Create a list of tuples with molecule and HBD count
molecule_with_hbd = [(mol, f"HBD: {hbd}") for mol, hbd in zip(molecules, hbd_counts)]

# Draw the molecules with HBD counts as captions
img = Draw.MolsToGridImage(
    [mol for mol, _ in molecule_with_hbd],
    legends=[hbd for _, hbd in molecule_with_hbd],
    molsPerRow=4,
    subImgSize=(200, 200)
)

# Display the image
img
```

```python
luna18 = "N3(C)[C@@H](C1CCCC1)C(=O)|N(C)[C@H](C(=O)N(C)C)CC(=O)|N(C)[C@@H](CC(C)C)C(=O)|N[C@@H]([C@H](CC)C)C(=O)|N(C)[C@@H](C)C(=O)|N1[C@@H](CC1)C(=O)|N(CC)[C@@H](Cc1ccc(C)cc1)C(=O)|N(C)CC(=O)|N[C@@H](CCc1cc(F)c(C(F)(F)F)c(F)c1)C(=O)|N1[C@@H](CCC1)C(=O)|NC1(CCCC1)C3(=O)"
luna18_peptide = ''.join(luna18.split('|'))
```

```python
from rdkit.Chem import Crippen

# Convert luna18_peptide to an RDKit molecule
luna18_mol = Chem.MolFromSmiles(luna18_peptide)

# Calculate the logP value
luna18_logp = Crippen.MolLogP(luna18_mol)

# Draw the molecule with logP as the caption
luna18_img = Draw.MolToImage(luna18_mol, legend=f"logP: {luna18_logp:.2f}")

# Display the image
luna18_img
```

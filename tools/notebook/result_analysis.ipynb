```python
# Define filtering conditions for different thresholds
conditions = [
    {"SOLY4": 120, "RatMRT": 2, "TNFR1-TRFRET-B372-V1": 1.0},
    {"SOLY4": 125, "RatMRT": 2.25, "TNFR1-TRFRET-B372-V1": 0.8},
    {"SOLY4": 130, "RatMRT": 2.5, "TNFR1-TRFRET-B372-V1": 0.6},
    {"SOLY4": 135, "RatMRT": 2.75, "TNFR1-TRFRET-B372-V1": 0.4},
    {"SOLY4": 140, "RatMRT": 3, "TNFR1-TRFRET-B372-V1": 0.3}
]

# Apply filters for each condition
for i, condition in enumerate(conditions):
    # Filter pepinvent based on conditions
    filtered_pepinvent = pepinvent[
        (pepinvent["raw_SOLY_4"] > condition["SOLY4"]) & 
        (pepinvent["raw_Rat_MRT"] > condition["RatMRT"]) & 
        (pepinvent["raw_TNFR1-TRFRET-B372-V1"] < condition["TNFR1-TRFRET-B372-V1"])
    ]
    
    # Filter pepinvent_pretrained based on conditions
    filtered_pepinvent_pretrained = pepinvent_pretrained[
        (pepinvent_pretrained["raw_SOLY_4"] > condition["SOLY4"]) & 
        (pepinvent_pretrained["raw_Rat_MRT"] > condition["RatMRT"]) & 
        (pepinvent_pretrained["raw_TNFR1-TRFRET-B372-V1"] < condition["TNFR1-TRFRET-B372-V1"])
    ]
    
    # Filter pepevolve_pretrained based on conditions
    filtered_pepevolve_pretrained = pepevolve_pretrained[
        (pepevolve_pretrained["raw_SOLY_4"] > condition["SOLY4"]) & 
        (pepevolve_pretrained["raw_Rat_MRT"] > condition["RatMRT"]) & 
        (pepevolve_pretrained["raw_TNFR1-TRFRET-B372-V1"] < condition["TNFR1-TRFRET-B372-V1"])
    ]
    
    # Print conditions and results
    print(f'Condition: SOLY4 > {condition["SOLY4"]}, RatMRT > {condition["RatMRT"]}, '
          f'TNFR1-TRFRET-B372-V1 < {condition["TNFR1-TRFRET-B372-V1"]}')
    print(f'Results: filtered_pepinvent: {len(filtered_pepinvent)}, '
          f'filtered_pepinvent_pretrained: {len(filtered_pepinvent_pretrained)}, '
          f'filtered_pepevolve_pretrained: {len(filtered_pepevolve_pretrained)}')
    print('-' * 80)  # Separator line for better readability
```
